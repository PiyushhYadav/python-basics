### Ceil()
ceil(x) will return the ceiling value of number x which we pass in argument. The ceiling value of a number x will be the smallest integer not less than x.
NOTE- This function will not be accessible directly using ceil() method. Math module will be required to access this method.

```python
import math
math.ceil(43.67)
```

```python
math.ceil(0.24)
```

```python
math.ceil(-45.7)

#In negative numbers, the ceiling value will be one less than its original value.
```

## fabs: Retuns absolute value but in floats
## abs() vs math.fabs()

- `abs()` → built-in, works with int, float, complex, returns appropriate type  
- `math.fabs()` → from math module, works with real numbers only, always returns float  

Example:
- `abs(-5)` → 5  
- `math.fabs(-5)` → 5.0
